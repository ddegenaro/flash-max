```python
import pandas as pd
import numpy as np
import os
import glob
```

```python
dfs = {
    os.path.basename(x).split('_')[0]: pd.read_csv(x, sep='\t')
    for x in glob.glob(os.path.join('fem_long_data', '*.tsv'))
}
```

```python
display_results = pd.DataFrame(columns=[
    'Solution', 'w/ BC', 'Min. RL2E. (%)', 'Time to <5% RL2E. and 0.01 PDE residual (s)', 'Time to min. RL2E. (s)',
    'Residual (at 5% err)', 'Residual (at min err)'
])

for name in dfs:
    df = dfs[name]
    try:
        first_row = df[(df['rel_l2_val_error'] < 0.05) & (df['pde_residual'] < 0.01)].iloc[0]
        time_to_5p = first_row['training_time'].item()
        residual_5p = first_row['pde_residual'].item()
    except:
        time_to_5p = np.nan
        residual_5p = np.nan
    min_err = df['rel_l2_val_error'].min() * 100.0
    time_to_min_err = df[df['rel_l2_val_error'] <= df['rel_l2_val_error'].min()].iloc[0]['training_time'].item()
    residual_min_err = df[df['rel_l2_val_error'] <= df['rel_l2_val_error'].min()].iloc[0]['pde_residual'].item()
    info = {
        'Solution': name,
        'w/ BC': True,
        'Min. RL2E. (%)': min_err,
        'Time to <5% RL2E. and 0.01 PDE residual (s)': time_to_5p,
        'Time to min. RL2E. (s)': time_to_min_err,
        'Residual (at 5% err)': residual_5p,
        'Residual (at min err)': residual_min_err
    }
    display_results.loc[len(display_results)] = info
    
display_results['Solution'] = display_results['Solution'].map({'rw': 4, 'hopf': 3, 'sp': 2, 'pw': 1})
display_results = display_results.sort_values(by='Solution').reset_index(drop=True)

display_results
```

```python
print(display_results.to_markdown(index=False))
```
