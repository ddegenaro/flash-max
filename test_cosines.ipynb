```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
df = pd.read_csv('experiments/907/log.tsv', sep='\t')
```

```python
df
```

```python
plt.figure(0)
time = df['training_time'].cumsum()
plt.plot(
    time,
    df['mse_train'],
    label='train'
)
plt.plot(
    time,
    df['mse_val'],
    label='val'
)
plt.xlabel('Time (s)')
plt.ylabel('MSE')
plt.legend(loc='upper right')
plt.title('0, cos(y+z), -cos(y+z), 0, 0, 0')
```
